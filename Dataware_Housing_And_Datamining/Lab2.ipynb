#Define Data Quality:

Data Quality refers to how accurate, reliable, and useful a set of data is for its intended purpose.

In simple terms, high-quality data means the data is correct, complete, consistent, and up-to-date, so it can be trusted for decision-making.

*****

#Issue with Data Quality: (In Data Mining & Data Warehousing):

1. Incomplete Data – Missing values or fields
2. Noisy Data – Errors, outliers, or random variations
3. Inconsistent Data – Conflicting information across sources
4. Duplicate Data – Same records repeated multiple times
5. Outdated Data – Data not updated regularly
6. Data Integration Issues – Mismatch from multiple sources
7. Data Entry Errors – Mistakes during data input

*****

#Missing Values and Its Reasons:

1. Data not collected – Some fields are left blank during data collection
2. Human errors – Mistakes while entering data
3. Equipment/system failure – Sensors or systems fail to record data
4. Data corruption – Data lost due to technical issues
5. Non-response – Users choose not to provide certain information.

******

#Techniques to Deal with Missing Values:
1. Ignore/Delete records – Remove rows with missing values (if few)
2. Manual filling – Fill missing values by hand (not practical for large data)
3. Use global constant – Replace with a value like “Unknown” or 0
4. Mean/Median/Mode substitution – Fill using average or most frequent value
5. Use attribute mean (class-based) – Mean calculated within same category/class
6. Prediction methods – Use algorithms (like regression, decision trees) to estimate

##5. Eliminating Dubllicated Rows:
