#LAB 4: Attribute Transformation and Dimensionality Reduction

##Objective

To apply attribute transformation and dimensionality reduction techniques on the Heart Disease dataset to improve data quality, reduce complexity, and prepare the dataset for data mining tasks.

###Definition of Attribute Transformation

Attribute transformation is the process of changing data into suitable formats for analysis or mining. It improves consistency, removes scale differences, and makes algorithms perform better.

Importance:

a. Makes data easier to analyze

b. Improves machine learning accuracy

c. Removes scale differences

d. Reduces skewness

e. Helps in feature engineering

###Definition of Dimensionality Reduction

Dimensionality reduction is the process of reducing the number of attributes while preserving important information.

Importance:

a. Reduces computation cost

b. Removes redundant features

c. Improves visualization

d. Helps reduce overfitting

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.decomposition import PCA
```

```python
heart_ds = pd.read_csv("/content/clean_heart_dataset.csv")
heart_ds.head()
```

```python
heart_ds.shape
```

```python
heart_ds.info()
```

```python
heart_ds.describe()
```

##ATTRIBUTE TRANSFORMATION

##1. Normalization (Min-Max Scaling)
*************
Definition

Normalization scales data values between 0 and 1.

Formula:

xnew = (x - xmin) / (xmax - xmin)


Why for this dataset?

features like:

1. cholesterol (chol)
2. age (age)
3. blood pressure (trestbps)

have different ranges. Scaling helps compare them.

```python
scaler = MinMaxScaler()

num_cols = ['age', 'trestbps', 'chol', 'thalach', 'oldpeak']

heart_ds[num_cols] = scaler.fit_transform(heart_ds[num_cols])

heart_ds.head()
```

##2. Standardization (Z-score)
***************
###Definition:

Transforms data to mean = 0 and standard deviation = 1.

Formula
Z=
(x−μ)/σ
	​

Why?

Useful when values vary a lot.

```python
std = StandardScaler()

heart_std = pd.DataFrame(
    std.fit_transform(heart_ds[num_cols]),
    columns=num_cols
)

heart_std.head()
```

###3. Log Transformation
*************************
###Definition

Log transformation reduces skewness.

Cholesterol and blood pressure can be right-skewed.

```python
log_df = heart_ds.copy()

log_df['chol'] = np.log1p(log_df['chol'])
log_df['trestbps'] = np.log1p(log_df['trestbps'])

log_df[['chol','trestbps']].head()
```

##4. Binning
*********************
###Definition:

Binning converts continuous values into categories.

Age can be grouped into patient age groups.

```python
heart_ds['age_group'] = pd.cut(
    heart_ds['age'],
    bins=3,
    labels=['Young','Middle','Old']
)

heart_ds[['age','age_group']].head()
```

##DIMENSIONALITY REDUCTION:

Dimensionality reduction is all about summarizing the data with most of the information preserved in compact form.Reducing the dimension of the feature space, creates fewer relationships between variables. PCA reduces features while keeping maximum information.

It is needed for:

a. Reduce number of attributes

b. Remove redundant features

c. Improve efficiency of analysis



###Step 1: Separate Features and Target:

Before applying dimensionality reduction, we separate:

i. Independent variables (X)

ii. Dependent variable (y)

```python
X = heart_ds.drop('target', axis=1)
y = heart_ds['target']

X.head()
```

###Step 2: Select Numerical Features

PCA works only on numerical data, so we select only numeric columns.

```python
X = X.select_dtypes(include=['int64', 'float64'])
X.head()
```

###Step 3: Feature Scaling (Standardization)

Scaling is necessary because:

a. Features have different ranges

b. PCA is sensitive to scale

We use Z-score standardization.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

###Step 4: Apply PCA

PCA converts original features into new features called:
Principal Components (PC1, PC2, etc.)

We reduce dataset into 2 components for simplicity.

```python
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)
```

###Step 5: Create PCA DataFrame

Store transformed data into a new dataset.

```python
pca_df = pd.DataFrame(X_pca, columns=['PC1', 'PC2'])
pca_df.head()
```

###Step 6: Check Explained Variance

Explained variance shows how much information is retained by each component.

```python
pca.explained_variance_ratio_
```

```python
print(pca.components_)
```

#Conclusion
********************
Dimensionality reduction was successfully performed using PCA. The original dataset with multiple features was reduced to two principal components while preserving most of the variance. This reduces complexity and improves efficiency for further analysis.
